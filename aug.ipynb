```python
S='kabab is love'.split()
out={}
for i in S :
    rev='' 
    vowel=0
    Char=''
    for j in i :
        if j in 'AEIOUaeiou':
            vowel+=1
    rev+=i[::-1]
    Char+=i[::2]
    out[i]=[rev,vowel,Char]
print(out)
        
         
```

```python
S='kabab is love'.split()
out={}
for i in S :
    const='' 
    consonat=0
    for j in i :
        if j not in 'AEIOUaeiou':
            consonat+=1
            const+=j
    out[i]=(const,consonat,const[::-1])
print(out)
        
```

```python
s = "bacbcaabbaa"
out = ""
for i in range(len(s)):
    count = 0
    flag = False
    for k in range(i):
        if s[i] == s[k]:
            flag = True
            break
    if flag:
        continue
    for j in range(len(s)):
        if s[i] == s[j]:
            count += 1
    out = out + s[i] + str(count)
print(out)
```

```python
L = [100, 200, 50, 400, 300]
n = 300
out = []
for i in range(len(L)):
    for j in range(i + 1, len(L)):
        if L[i] + L[j] == n:
            out.append([L[i], L[j]])
    if L[i] == n:
        out.append([L[i]])
print(out)                          
```
