Q4)WAP to reverse the string without the slicing

```python
Str=input('Enter your string : ')
Str1=''
for i in range(len(Str)-1,-1,-1):
    Str1=Str1+Str[i]
print(Str1)    
```

Q5) WAP to extract all the lowercase characters in a string only if the ascii value is even

```python
Str=input('Enter the String : ')
for i in Str :
    if i.islower() and ord(i)%2==0:
        print(i)
```

Q8) WAP to extract all the key value pair from the dictionary only if keys are of string datatpye and values are integers

```python
Dict={'Tarun':10,'Hi':100,10:'Tarun',22:'bye'}
for key , value in Dict.items():
    if type(key)==str and type(value)==int:
        print(key , value)
```

Q9) WAP to extract key value pair from the dict only if both keys and values are exactly same

```python
Dict={'Tarun':'Tarun','Hi':100,10:'Tarun',22:22}
for key , value in Dict.items():
    if key==value:
        print(key , value)
```
