Q12) WAP to check wheather the list is homogeneous or not 

```python
List=[10,10,20,29]
count=1
count1=0
for i in range(len(List)+1):
    if type(List[0])==type(List[i]):
        count+=1
    else:
        count1=1 
        break 
if count==len(List):
    print('This is Homogeneous')
else:
    print('This is not homogeneous')        
```

```python
#WAP to toggle a string 
Str=input('Enter the string : ')
Str1=''
for i in Str :
    if 'a'<=i<='z':
        Str1+=i.upper() 
    elif i==' ':
        Str1+=' '
    elif 'A'<=i<='Z':
        Str1+=i.lower()
    else:
        continue 
print(Str1)
```

```python
#WAP extract upper , lower , digit and special characters present in a string to different output variable 
Str=input('Enter the String : ')
Upper=''
lower=''
digit=''
special_Char=''
for i in Str :
    if i.islower():
        lower+=i 
    elif i.isupper():
        Upper+=i 
    elif i.isdigit():
        digit+=i 
    else:
        special_Char+=i
print('The Uppercase present in the string are :', Upper)
print('The lowercase present in the string are : ',lower)
print( 'The Digit present in the string are :' ,digit)
print( 'The Special Character present in the string are :' , special_Char)
```
