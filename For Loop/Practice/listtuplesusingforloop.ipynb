Q1) Find the largest element in a list.

```python
L = [12, 45, 7, 89, 23, 56]
largest = L[0]
for i in L:
    if i > largest:
        largest = i
print("Largest Element:", largest)
```

Q2) Find the smallest element in a list.

```python
L = [12, 45, 7, 89, 23, 56]
Smallest = L[0]
for i in L:
    if i < Smallest :
        Smallest  = i
print("Smallest Element:", Smallest)
```

Q3) Find the second largest element in a list.

```python
L = [12, 45, 7, 89, 23, 56]
largest = L[0]
secondlargest = L[0]
for i in L:
    if i > largest:
        secondlargest = largest
        largest = i
    elif i > secondlargest and i != largest:
        secondlargest = i
print("Largest:", largest)
print("Second Largest:", secondlargest)
```

Q4)Count how many even and odd numbers are in a list.

```python
L=[10,11,12,13,14,15,16,17,18]
even_Count=0 
odd_Count=0
for i in range(len(L)):
    if i%2==0:
        even_Count+=1 
    else :
        odd_Count+=1 
print('Total Number of the Even Number in the list is : ',even_Count)
print('Total Number of the odd Numbers in the list is : ',odd_Count)
```

Q5) Find the sum of all list elements.

```python
L=[11,23,44,445,543]
Sum=0
for i in L :
    Sum+=i 
print('Sum of the numbers present in the list is : ',Sum)
    
```

Q6) Print only string values from a mixed list.

```python
L=['Tarun',21,'Hi',22,'Bye']
print('The Strings values present in the mixed list is as follows : ')
for i in L:
    if type(i)==str:
        print(i)
```

Q7)Count how many palindrome strings are in a list.

```python
L=[11,'abba','abccba','Tarun',78,90]
Count=0
for i in L :
    if type(i)==str and i==i[::-1]:
        Count+=1
print('Total Number of palindrome string present in the List are : ',Count)
        
```

Q8) Find the longest string in a list.

```python
L=['Tarun','Badh','Hiiiiiiii','Byeeee!!!!!']
Longest_String=L[0]
for i in L:
    if len(i)>len(Longest_String):
        Longest_String=i 
print('The longest string for the list is : ',Longest_String)
```

Q9) Remove duplicates without using set().

```python
List=[10,10,20,20,80,90,80,70]
L=[]
for i in List :
    if i not in L:
        L.append(i)
print(L)
```
