Q1) Count the number of vowels in a string.

```python
#Count the number of vowels in a string.
#Input : tarun
Str = input('Enter your String to check number of vowels : ')
count=0
for i in Str :
    if i in 'AEIOUaeiou':
        count+=1 
print('Total Number of vowels in string is : ', count)
```

Q2)Count the number of consonants in a string.

```python
#Count the number of consonants in a string.
#input : tarun
Str = input('Enter your String to check number of consonants : ')
count=0
for i in Str :
    if i not in 'AEIOUaeiou':
        count+=1 
print('Total Number of consonants in string is : ', count)
```

Q3)Reverse a string using only a for loop. 

```python
#Reverse a string using only a for loop. 
Str = input('Enter your String which you want to reverse : ')
rev_Str=''
for i in range(len(Str)-1,-1,-1) :
    rev_Str=rev_Str+Str[i]
print('The reversed string is : ',rev_Str)
```

Q4)Reverse a string using only a for loop. 

```python

# Ip: tarun , abba
s = input("Enter a String: ")
rev = ""
for i in range(len(s)-1, -1, -1):
    rev = rev + s[i]
if s == rev:
    print("Palindrome")
else:
    print("Not Palindrome")
```

Q5) Count the number of uppercase and lowercase letters.

```python
Str=input('Enter the String : ')
no_of_Uppercase = 0 
no_of_Lowercase = 0 
for i in range(len(Str)):
    if 'a'<=Str[i]<='z':
        no_of_Lowercase+=1 
    elif 'A'<=Str[i]<='Z':
        no_of_Uppercase+=1 
    else :
        print(Str[i],' is a special character you have used in the string or digit ')
print('Total number of the uppercase in the string is : ',no_of_Uppercase)
print('Total number of the lowercase in the string is : ',no_of_Lowercase)
```

Q6)Find the sum of digits of a number using a for loop.

```python
#IP : 234
Num=int(input('Enter the Number : '))
sum=0 
for i in str(Num):
    sum=sum+int(i)
print('Sum of Digits : ',sum)
    
```

Q8) Check whether a number is prime

```python
#IP : 7
Num = int(input('Enter the Number: '))
count = 0
for i in range(2, Num):
    if Num % i == 0:
        count += 1
if count == 0 and Num > 1:
    print(Num, "is a prime number")
else:
    print(Num, "is not a prime number")
        
```

Q9) Print all prime numbers from 1 to n

```python
#IP : 88
n = int(input("Enter the Number: "))
for num in range(2, n + 1):
    count = 0
    for i in range(2, num):
        if num % i == 0:
            count += 1
    if count == 0:
        print(num)
```

Q10) Print Fibonacci series using a for loop.

```python
#IP : 9
Num = int(input("Enter the Number: "))
a = 0
b = 1
print("The Fibonacci Series for the given number is:")
for i in range(Num):
    print(a, end=", ")
    c = a + b
    a = b
    b = c
    
```
